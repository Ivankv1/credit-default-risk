```python
key_num_features = [
    'AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY', 'AMT_GOODS_PRICE',
    'DAYS_BIRTH', 'DAYS_EMPLOYED', 'DAYS_REGISTRATION', 'DAYS_ID_PUBLISH',
    'CNT_CHILDREN', 'CNT_FAM_MEMBERS', 'REGION_POPULATION_RELATIVE',
    'OWN_CAR_AGE', 'EXT_SOURCE_1', 'EXT_SOURCE_2', 'EXT_SOURCE_3'
]
print('Статистика по ключевым числовым признакам:')
display(train[key_num_features].describe().T.round(2))
print('Пропуски в ключевых признаках:')
train[key_num_features].isnull().mean().sort_values(ascending=False).round(3)
```

```python
def target_by_category(df, col):
  temp = df.groupby(col)['TARGET'].agg(['mean', 'count']).sort_values('mean', ascending=False)
  temp['mean'] = (temp['mean'] * 100).round(2)
  temp.columns = ['Доля дефолта, %', 'Количество']
  return temp

print("=== CODE_GENDER ===")
display(target_by_category(train, 'CODE_GENDER'))

print("\n=== NAME_INCOME_TYPE ===")
display(target_by_category(train, 'NAME_INCOME_TYPE'))

print("\n=== NAME_EDUCATION_TYPE ===")
display(target_by_category(train, 'NAME_EDUCATION_TYPE'))

print("\n=== NAME_FAMILY_STATUS ===")
display(target_by_category(train, 'NAME_FAMILY_STATUS'))

print("\n=== FLAG_OWN_CAR ===")
display(target_by_category(train, 'FLAG_OWN_CAR'))
```
